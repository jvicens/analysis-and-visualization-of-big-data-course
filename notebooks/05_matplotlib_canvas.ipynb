## 1. Figure and Axes

A **Figure** is the canvas; an **Axes** is one plot with its own x and y axis. Everything drawn is an
*Artist*. Always keep a handle on both:

```python
x = np.linspace(0, 2 * np.pi, 200)
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, np.sin(x), label="sin")
ax.plot(x, np.cos(x), label="cos")
ax.set_xlabel("x (rad)")
ax.set_ylabel("value")
ax.legend()
fig.suptitle("Figure title (owned by the Figure)")
plt.show()
```

## 2. Grids of panels

`plt.subplots(nrows, ncols)` returns an array of Axes. `layout="constrained"` spaces them so nothing
overlaps (the modern replacement for `tight_layout()`).

```python
fig, axs = plt.subplots(2, 3, figsize=(9, 4), layout="constrained")
for i, ax in enumerate(axs.flat, start=1):
    ax.plot(x, np.sin(i * x))
    ax.set_title(f"sin({i}x)")
plt.show()
```

## 3. Mosaics

`subplot_mosaic` describes the layout as a picture: the same letter spans several cells.

```python
fig, axd = plt.subplot_mosaic("AAB;CCB", figsize=(8, 4), layout="constrained")
for name, ax in axd.items():
    ax.text(0.5, 0.5, name, ha="center", va="center", fontsize=24)
plt.show()
```

## 4. Insets, projections, shared axes

```python
fig, ax = plt.subplots(figsize=(5, 3.5))
ax.plot(x, np.sin(x) * np.exp(-x / 3))
inset = ax.inset_axes([0.55, 0.5, 0.4, 0.4])     # [x0, y0, width, height] in axes fraction
inset.plot(x[:40], np.sin(x[:40]) * np.exp(-x[:40] / 3), color="tab:red")
inset.set_title("zoom", fontsize=8)
plt.show()
```

```python
fig, axs = plt.subplots(1, 2, figsize=(8, 3.5), subplot_kw={"projection": "polar"}, layout="constrained")
theta = np.linspace(0, 2 * np.pi, 400)
axs[0].plot(theta, 1 + 0.5 * np.cos(5 * theta))
axs[1].plot(theta, np.abs(np.sin(2 * theta)))
plt.show()
```

```python
fig, axs = plt.subplots(3, 1, figsize=(5, 5), sharex=True, layout="constrained")
for k, ax in enumerate(axs, start=1):
    ax.plot(x, np.sin(k * x))
axs[-1].set_xlabel("x (rad)")     # shared x: label only the bottom panel
plt.show()
```
